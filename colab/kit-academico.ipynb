# Kit de escritura académica: de Markdown a Word

Para cuando tu IA (Gemini, Perplexity, Copilot, Grok, DeepSeek…) te dio el trabajo en **Markdown** con el encabezado del kit y no puede crear el Word.

1. **Entorno de ejecución › Ejecutar todo** (o el botón ▶ de cada celda, en orden).
2. En la celda 2, subí tu archivo `.md` (o pegá el texto en `TEXTO`). También podés subir un `.docx` que ya te dio la IA: se revisa y se verifican sus fuentes, que el Python de los chats no puede hacer porque no tiene internet.
3. Al final se descarga el `.docx` con formato APA 7 y ves la revisión del kit.

No se guarda nada fuera de tu sesión de Colab. Instrucciones del kit: https://github.com/josueazc/academic-writing-kit

```python
# 1. Descargar el kit
import os, subprocess, sys
if not os.path.isdir("kit"):
    subprocess.run(["git", "clone", "-q", "--depth", "1", "https://github.com/josueazc/academic-writing-kit", "kit"], check=True)
print("Kit listo:", subprocess.run(["git", "-C", "kit", "log", "-1", "--format=%h %cd", "--date=short"], capture_output=True, text=True).stdout)
```

```python
# 2. Tu trabajo: subí el .md (aparece un botón) o pegá el texto entre las comillas de TEXTO
TEXTO = """"""

ARCHIVO = "trabajo.md"
if TEXTO.strip():
    open(ARCHIVO, "w", encoding="utf-8").write(TEXTO.strip() + "\n")
else:
    try:
        from google.colab import files
        subidos = files.upload()
        ARCHIVO = next(n for n in subidos if n.lower().endswith((".md", ".txt", ".markdown", ".docx")))
    except StopIteration:
        raise SystemExit("Subí un archivo .md o .docx (o pegá el texto en TEXTO y volvé a ejecutar).")
print("Trabajo:", ARCHIVO)
```

```python
# 3. Revisar: ¿existen las fuentes? ¿citas, objetivos y formato bien? (Colab sí tiene internet)
for script in ("verificar_referencias.py", "revisar.py"):
    r = subprocess.run([sys.executable, f"kit/scripts/{script}", ARCHIVO], capture_output=True, text=True)
    print(r.stdout or r.stderr)
```

```python
# 4. Generar el Word y descargarlo (si subiste un .docx, este paso no hace falta)
if ARCHIVO.lower().endswith(".docx"):
    raise SystemExit("Subiste un .docx: ya está generado. Corregí lo que marcó la revisión.")
r = subprocess.run([sys.executable, "kit/scripts/md2docx.py", ARCHIVO], capture_output=True, text=True)
print(r.stdout or r.stderr)
docx = os.path.splitext(ARCHIVO)[0] + ".docx"
try:
    from google.colab import files
    files.download(docx)
except ImportError:
    print("Archivo:", os.path.abspath(docx))
print("Al abrirlo en Word, aceptá actualizar los campos para que el índice muestre los números de página.")
```
